```python
num=[70,80,60,30,25,12,100,50]
def above(num):
    count=0
    for n in num:
        if n>50:
            count += 1
    return count
print("Numbers above 50: ",above(num))
```

A small menu using match case

```python
items = []
while True:
    print("1. Add")
    print("2. View")
    print("3. Exit")
    choice = input("Choose an option: ").strip()

    match choice:
        case "1":
            item = input("Enter item: ")
            items.append(item)
            print("Added")
        case "2":
            if not items:
                print("Nothing to show yet.")
            else:
                for i, item in enumerate(items, 1):
                    print(f"{i}. {item}")
        case "3":
            print("Goodbye")
            break
        case _ :
            print("Invalid choice, try again.")
```
